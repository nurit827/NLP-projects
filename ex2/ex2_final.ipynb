# Ex2 — NLP 67658

```python
# Shared imports + reproducibility + hyperparameters
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

from ex2 import MLP_classification, transformer_classification, category_dict

seed = 42
np.random.seed(seed)
torch.manual_seed(seed)

feature_dim = 2000
hidden_dim = 500
num_labels = len(category_dict)

class MLP(nn.Module):
    def __init__(self, input_dim, hidden_dim, num_labels):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, num_labels),
        )
    def forward(self, x):
        return self.net(x)

print('Setup done.')
```

# Q1 — Log-linear classifier (TFIDF + single-layer perceptron)

## Q1 (a) portion=0.1

```python
# Train log-linear (portion=0.1) — delegates to ex2.MLP_classification
q1_train_losses_a, q1_val_accs_a = MLP_classification(
    portion=0.1, model=nn.Linear(feature_dim, num_labels))
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(range(1, len(q1_train_losses_a) + 1), q1_train_losses_a, marker='o')
ax.set_xlabel('Epoch'); ax.set_ylabel('Train loss')
ax.set_title('Q1 Log-linear (portion=0.1) — Train loss')
ax.xaxis.set_major_locator(MaxNLocator(integer=True))
plt.tight_layout(); plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(range(1, len(q1_val_accs_a) + 1), q1_val_accs_a, marker='o', color='tab:orange')
ax.set_xlabel('Epoch'); ax.set_ylabel('Validation accuracy')
ax.set_title('Q1 Log-linear (portion=0.1) — Val accuracy')
ax.xaxis.set_major_locator(MaxNLocator(integer=True))
plt.tight_layout(); plt.show()
```

The exact epoch of the peak shifts between portions and even between random seeds, which is itself a sign of high sensitivity.
                                                                                                                                                                                                                                                 
  The MLP was moderately sensitive: it overfits more aggressively at larger portions but the curve shapes do not move around as much.                                                                                                            
   
  The log-linear classifier was the least sensitive: accuracy improves slowly with more data and the curves look similar across portions.                                                                                                        
                  
  Order: transformer > MLP > log-linear.                                                                                                                                                                                                         
                  


# c

**Trainable parameters (extracted via `sum(p.numel() for p in model.parameters() if p.requires_grad)`):**                                                                                                                                      
                                                                                                                                                                                                                                               
  | Model | Trainable parameters |                                                                                                                                                                                                               
  |---|---:|                                                                                                                                                                                                                                     
  | Q1 (linear): (`Linear(2000, 4)`) | 8,004 |                                                                                                                                                                                                
  | Q2 (MLP): (`Linear(2000, 500) → ReLU → Linear(500, 4)`) | 1,002,504 |                                                                                                                                                                         
  | Q3  (transformer + classification head):| 82,121,476 |                                                                                                                                                                                 
                                                                                                                                                                                                                                                 
  **Do additional parameters help?**                                                                                                                                                                                                             
                                                                                                                                                                                                                                                 
  Yes (given enough data) but they bring an overfitting cost
                                                                                                                                                                                                                                                 
  - More parameters give the model more capacity to capture useful patterns, which is why the transformer (~82M) beats the MLP (~1M), which beats the log-linear (~8K) when each is given enough training examples.
  - The same extra capacity makes the model more prone to overfitting when data is scarce: the transformer at `portion=0.1` overfits within 2 epochs, while the log-linear barely overfits at all on the same data. On the same note, we can see MLP model overfits much more than the log-linear single layer and much less than the transformers on the comparable epoch and dataset size                               
                                                                                                                                                                                                                                                 
  In short: **more parameters → higher peak accuracy, but higher sensitivity to training-set size.**
